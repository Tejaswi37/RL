```python
n = int(input("Enter number of machines: "))

machines = []
q_values = []

for i in range(n):
    name = input(f"\nEnter machine name {i + 1}: ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))

    q = sum(rewards) / len(rewards)

    machines.append((name, rewards))
    q_values.append(q)

    print(f"\nMachine {name}")
    print(f"Rewards: {[int(r) if r.is_integer() else r for r in rewards]}")
    print(f"Q({name}) = {q:.2f}")

best_index = q_values.index(max(q_values))

print(f"\nGreedy Selected Machine: {machines[best_index][0]}")
```

```python
machines = {
    "A": [3, 2, 4],
    "B": [6, 7],
    "C": [5, 4, 6]
}

Q = {"A": 0, "B": 0, "C": 0}
N = {"A": 0, "B": 0, "C": 0}

for machine, rewards in machines.items():
    for R in rewards:
        N[machine] += 1
        Q[machine] = Q[machine] + (R - Q[machine]) / N[machine]

        best = max(Q, key=Q.get)

        print(f"Machine {machine}, Reward = {R}")
        print(f"N = {N[machine]}, Q = {Q[machine]:.2f}")
        print(f"Greedy Machine = {best}\n")

print("Final Q values:", Q)
print("Final Greedy Selected Machine:", max(Q, key=Q.get))
```
