```python
import gymnasium as gym


def random_agent(env_name):

    # Create environment
    env = gym.make(env_name)

    # Reset environment
    state, info = env.reset()

    total_reward = 0
    steps = 0

    print("\n======================================")
    print("Environment:", env_name)
    print("======================================")

    print("Initial State:", state)
    print("Action Space:", env.action_space)

    while True:

        # Select random action
        action = env.action_space.sample()

        # Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        # Display interaction
        print("Step:", steps + 1,
              "| Action:", action,
              "| Reward:", reward,
              "| Next State:", next_state)

        # Update values
        state = next_state
        total_reward += reward
        steps += 1

        # Stop when episode ends
        if terminated or truncated:
            break

    # Calculate average reward
    average_reward = total_reward / steps

    print("\nTotal Reward:", total_reward)
    print("Number of Steps:", steps)
    print("Average Reward:", round(average_reward, 2))

    env.close()


# Run all three environments
random_agent("CartPole-v1")
random_agent("MountainCar-v0")
random_agent("FrozenLake-v1")
```
