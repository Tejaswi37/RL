```python
import random

# ============================================================
# Part A – Machine Initialization
# ============================================================

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# N = number of times each machine is selected
N = [0, 0, 0]

# R = total reward received from each machine
R = [0, 0, 0]

# Q = estimated average reward of each machine
Q = [0.0, 0.0, 0.0]

# Reward patterns given in the question
reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of which reward to use next
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Fixed seed so that the experiment is reproducible
random.seed(42)

total_reward = 0

print("=" * 70)
print("       MODIFIED EPSILON-GREEDY MULTI-ARMED BANDIT")
print("=" * 70)

# ============================================================
# Part B, C, D and E – Decision, Reward and Q-value Update
# ============================================================

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Find machine with highest Q-value
    highest_index = Q.index(max(Q))

    # --------------------------------------------------------
    # Modified Epsilon-Greedy Decision
    # --------------------------------------------------------

    if r < epsilon:
        # Exploration:
        # Select randomly from machines EXCEPT highest Q machine
        available_machines = [
            i for i in range(M)
            if i != highest_index
        ]

        selected_index = random.choice(available_machines)
        decision = "Exploration"

    else:
        # Exploitation:
        # Select machine with highest Q-value
        selected_index = highest_index
        decision = "Exploitation"

    selected_machine = machines[selected_index]

    # --------------------------------------------------------
    # Part D – Reward Collection
    # --------------------------------------------------------

    index = reward_index[selected_machine]

    reward = reward_pattern[selected_machine][index]

    # Move to next reward for this machine
    reward_index[selected_machine] += 1

    # --------------------------------------------------------
    # Part E – Incremental Sample-Average Update
    # --------------------------------------------------------

    # Update participation count
    N[selected_index] += 1

    # Update total reward
    R[selected_index] += reward

    # Incremental sample-average formula:
    #
    # Q_new = Q_old + (Reward - Q_old) / N
    #
    Q[selected_index] = (
        Q[selected_index]
        + (reward - Q[selected_index]) / N[selected_index]
    )

    total_reward += reward

    # --------------------------------------------------------
    # Display current play
    # --------------------------------------------------------

    print(f"\nPlay {play}")
    print("-" * 40)
    print(f"Random number : {r:.4f}")
    print(f"Decision      : {decision}")
    print(f"Selected      : Machine {selected_machine}")
    print(f"Reward        : {reward}")

    print("\nCurrent N values:")
    print(f"A = {N[0]}, B = {N[1]}, C = {N[2]}")

    print("Current R values:")
    print(f"A = {R[0]}, B = {R[1]}, C = {R[2]}")

    print("Current Q values:")
    print(f"A = {Q[0]:.2f}, B = {Q[1]:.2f}, C = {Q[2]:.2f}")


# ============================================================
# Part F – Final Results
# ============================================================

print("\n")
print("=" * 70)
print("                     FINAL RESULTS")
print("=" * 70)

print("\nParticipation Count (N)")
print("-" * 30)
print("Machine\tN")
print(f"A\t{N[0]}")
print(f"B\t{N[1]}")
print(f"C\t{N[2]}")

print("\nTotal Rewards (R)")
print("-" * 30)
print("Machine\tR")
print(f"A\t{R[0]}")
print(f"B\t{R[1]}")
print(f"C\t{R[2]}")

print("\nEstimated Q-values")
print("-" * 30)
print("Machine\tQ-value")
print(f"A\t{Q[0]:.2f}")
print(f"B\t{Q[1]:.2f}")
print(f"C\t{Q[2]:.2f}")

print("\nTotal Reward Received =", total_reward)

print("\nBest Machine =", machines[Q.index(max(Q))])

print("=" * 70)
```
