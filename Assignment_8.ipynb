```python
# Assignment 8
# Calculate State Value and Action Value
# Using trajectory method (without Bellman Equation)

# import grid_env_det


# ---------------------------------------------------
# 1. Environment
# ---------------------------------------------------

gamma = 0.9

# Given policy
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}


# ---------------------------------------------------
# 2. Grid World definition
# ---------------------------------------------------

# State transitions
# Format: state -> action -> (next_state, reward)

transitions = {
    "S0": {
        "Up":    ("S0", -1),
        "Down":  ("S0", -1),   # obstacle X
        "Left":  ("S0", -1),
        "Right": ("S1", -1)
    },

    "S1": {
        "Up":    ("S1", -1),
        "Down":  ("S3", -1),
        "Left":  ("S0", -1),
        "Right": ("G", 10)
    },

    "S3": {
        "Up":    ("S1", -1),
        "Down":  ("S3", -1),
        "Left":  ("S3", -1),
        "Right": ("S4", -1)
    },

    "S4": {
        "Up":    ("S1", -1),
        "Down":  ("S4", -1),
        "Left":  ("S3", -1),
        "Right": ("S4", -1)
    }
}


# ---------------------------------------------------
# 3. Calculate State Value using trajectory method
# ---------------------------------------------------

def state_value(start_state, max_steps=1000):

    state = start_state
    discounted_return = 0.0
    discount = 1.0

    visited = set()

    for step in range(max_steps):

        # Goal state
        if state == "G":
            break

        # Detect cycle
        key = state

        # If same state occurs again under same policy,
        # the policy has entered a loop.
        if key in visited:

            # All remaining rewards are -1.
            # Infinite discounted sum:
            #
            # -1 / (1 - gamma)

            discounted_return += discount * (-1 / (1 - gamma))
            break

        visited.add(key)

        # Select action according to policy
        action = policy[state]

        # Get next state and reward
        next_state, reward = transitions[state][action]

        # Discounted reward
        discounted_return += discount * reward

        # Update discount
        discount *= gamma

        # Move to next state
        state = next_state

    return discounted_return


# ---------------------------------------------------
# 4. Calculate Action Value using trajectory method
# ---------------------------------------------------

def action_value(start_state, action, max_steps=1000):

    # Take the requested first action
    next_state, reward = transitions[start_state][action]

    # First reward
    total_return = reward

    discount = gamma
    state = next_state

    visited = set()

    for step in range(max_steps):

        # Goal reached
        if state == "G":
            break

        # Detect cycle
        if state in visited:

            # Remaining rewards are -1
            total_return += discount * (-1 / (1 - gamma))
            break

        visited.add(state)

        # Follow the given policy
        policy_action = policy[state]

        next_state, reward = transitions[state][policy_action]

        # Add discounted reward
        total_return += discount * reward

        # Update discount
        discount *= gamma

        # Move to next state
        state = next_state

    return total_return


# ---------------------------------------------------
# 5. Display State Value for S0
# ---------------------------------------------------

V_S0 = state_value("S0")

print("========================================")
print("STATE VALUE")
print("========================================")
print("Policy:")
print(policy)

print("\nState Value of S0:")
print("V(S0) =", V_S0)


# ---------------------------------------------------
# 6. Display Action Value for S0 and Up
# ---------------------------------------------------

Q_S0_Up = action_value("S0", "Up")

print("\n========================================")
print("ACTION VALUE")
print("========================================")

print("Action Value of S0 with action Up:")
print("Q(S0, Up) =", Q_S0_Up)


# ---------------------------------------------------
# 7. Display values of all policy states
# ---------------------------------------------------

print("\n========================================")
print("STATE VALUES FOR ALL POLICY STATES")
print("========================================")

for state in ["S0", "S1", "S3", "S4"]:
    value = state_value(state)
    print(f"V({state}) = {value:.2f}")
```
